```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import uproot
import matplotlib
```

```python
def applyCut(inputDataframe, cut, text=None):
    dataframe = inputDataframe
    nbeforecut = dataframe.shape[0]
    cutDataframe = dataframe.query(cut)
    if text:
        print (text, cutDataframe.shape[0], ' fraction kept: %2.1f'%(100.0*float(cutDataframe.shape[0])/nbeforecut))
    return cutDataframe

def getData(observables, filename, treename):
    return (uproot.open(filename)[treename]).arrays(observables, library='pd')

def getRcAndErr(df, ob):
    if (ob == 'b'):
        nos = sum(df[df.b==-1]['w'])
        nss = sum(df[df.b==+1]['w'])
    elif (ob == 'b_ch'):
        nos = sum(df[df.b_ch==-1]['w'])
        nss = sum(df[df.b_ch==+1]['w'])
    elif (ob == 'bs'):
        nos = sum(df[df.bs==-1]['ws'])
        nss = sum(df[df.bs==+1]['ws'])
    else:
        print('Double check the input observable name!')
    rc = (nss-nos)/(nss+nos)
    p = nss/(nss+nos)
    w_array = np.unique(df['w'])
    err2_array = []
    variance_array_test = []
    for i in range(0, len(w_array)):
        if (ob == 'b'):
            p = sum(df[(df.w == w_array[i]) & (df.b==+1)]['w']) / sum(df[df.w == w_array[i]]['w'])
        elif (ob == 'b_ch'):
            p = sum(df[(df.w == w_array[i]) & (df.b_ch==+1)]['w']) / sum(df[df.w == w_array[i]]['w'])            
        elif (ob == 'bs'):
            p = sum(df[(df.w == w_array[i]) & (df.bs==+1)]['w']) / sum(df[df.w == w_array[i]]['ws'])
        else:
            print('Double check the input observable name!')        
        variance = p*(1-p)/df[df.w == w_array[i]].shape[0]
        w = sum(df[df.w == w_array[i]]['w'])/sum(df['w'])
        err2_array.append ( w**2*variance )
    rc_err = 2*np.sqrt(sum(err2_array))
    return rc, rc_err

def calc_hist(vals, bins=10, weights=None, density=False):    
    if weights is None:
        weights = np.ones(vals.shape)
    hist, bins = np.histogram(vals, bins=bins, weights=weights)
    digits = np.digitize(vals, bins)
    errs = np.asarray([np.linalg.norm(weights[digits==i]) for i in range(1, len(bins))])
    if density:
        density_int = weights.sum() * (bins[1] - bins[0])
        hist /= density_int
        errs /= density_int
    return hist, errs, bins
```

```python
observables1 = ['pt','b','dr','z','epair','w','reject','pt1','pt2','qlead','qsublead','y1','y2','phi1','phi2']
observabless1 = ['pts','bs','drs','zs','epairs','ws','rejects','pt1s','pt2s','qleads','qsubleads','y1s','y2s','phi1s','phi2s']
df_all = applyCut(getData(observables1+observabless1, '/gpfs/gibbs/project/havener/ys668/other_output/rc/all0615_chargedrc.root', "MatchedTree"),'reject<=0')
```

```python
distance1 = []
distance2 = []
for i in range(0,df_all.shape[0]):
    phi1small = min(df_all.iloc[i]['phi1'], df_all.iloc[i]['phi1s'])
    phi1large = max(df_all.iloc[i]['phi1'], df_all.iloc[i]['phi1s'])
    phi2small = min(df_all.iloc[i]['phi2'], df_all.iloc[i]['phi2s'])
    phi2large = max(df_all.iloc[i]['phi2'], df_all.iloc[i]['phi2s'])
    dphi1 = min(phi1large-phi1small, 6.283185307179586-(phi1large-phi1small))
    dphi2 = min(phi2large-phi2small, 6.283185307179586-(phi2large-phi2small))
    if (dphi1 > np.pi):
        print(df_all.iloc[i]['phi1'], df_all.iloc[i]['phi1s'], dphi1)
    distance1.append(np.sqrt(dphi1**2+(df_all.iloc[i]['y1']-df_all.iloc[i]['y1s'])**2))
    distance2.append(np.sqrt(dphi2**2+(df_all.iloc[i]['y2']-df_all.iloc[i]['y2s'])**2))
df_all['distance1'] = distance1
df_all['distance2'] = distance2
```

```python
df_alltruth = applyCut(df_all, 'pt>5 and b!=-9')
```

```python
df_match = applyCut(df_alltruth, 'pts>15 and bs!=-9 and distance1<0.02 and distance2<0.02')
```

```python
def get_eff3(df):
    true = np.histogram(df['pt'], bins=binedges, weights=df['w'])[0]
    match = np.histogram(df[(df.pts>15) & (df.bs!=-9)]['pt'], bins=binedges, weights=df[(df.pts>15) & (df.bs!=-9)]['w'])[0]
    return match/true
```

```python
def get_eff4(df):
    true = np.histogram(df['pt'], bins=binedges, weights=df['w'])[0]
    match = np.histogram(df[(df.pts>15) & (df.bs!=-9) & (df.distance1<0.02) & (df.distance2<0.02)]['pt'], bins=binedges, weights=df[(df.pts>15) & (df.bs!=-9) & (df.distance1<0.02) & (df.distance2<0.02)]['w'])[0]
    return match/true
```
