```python
def applyCut(inputDataframe, cut, text=None):
    dataframe = inputDataframe
    nbeforecut = dataframe.shape[0]
    cutDataframe = dataframe.query(cut)
    if text:
        print (text, cutDataframe.shape[0], ' fraction kept: %2.1f'%(100.0*float(cutDataframe.shape[0])/nbeforecut))
    return cutDataframe
```

```python
w0=(float)(df0.iloc[df0.shape[0]-1]['xsec'])
w1=(float)(df1.iloc[df1.shape[0]-1]['xsec'])
w2=(float)(df2.iloc[df2.shape[0]-1]['xsec'])
w3=(float)(df3.iloc[df3.shape[0]-1]['xsec'])
w4=(float)(df4.iloc[df4.shape[0]-1]['xsec'])
w5=(float)(df5.iloc[df5.shape[0]-1]['xsec'])
w6=(float)(df6.iloc[df6.shape[0]-1]['xsec'])
w7=(float)(df7.iloc[df7.shape[0]-1]['xsec'])
w8=(float)(df8.iloc[df8.shape[0]-1]['xsec'])
df0['w']=np.ones(df0.shape[0])*w0
df1['w']=np.ones(df1.shape[0])*w1
df2['w']=np.ones(df2.shape[0])*w2
df3['w']=np.ones(df3.shape[0])*w3
df4['w']=np.ones(df4.shape[0])*w4
df5['w']=np.ones(df5.shape[0])*w5
df6['w']=np.ones(df6.shape[0])*w6
df7['w']=np.ones(df7.shape[0])*w7
df8['w']=np.ones(df8.shape[0])*w8
df = pd.concat([df0,df1,df2,df3,df4,df5,df6,df7,df8])
df_pythia8 = df[['pt','b_ch','w','y1_ch', 'y2_ch', 'phi1_ch', 'phi2_ch']]
df_pythia8 = applyCut(df_pythia8, 'b_ch!=0 and b_ch!=-9 and pt>5')
```

```python
df_pythia8 = applyCut(df_pythia8, '20<pt<30')
```

```python
dr = []
for i in range(0,df_pythia8.shape[0]):
    phismall = min(df_pythia8.iloc[i]['phi1_ch'], df_pythia8.iloc[i]['phi2_ch'])
    philarge = max(df_pythia8.iloc[i]['phi1_ch'], df_pythia8.iloc[i]['phi2_ch'])
    dphi = min(philarge-phismall, 6.283185307179586-(philarge-phismall))
    if (dphi > np.pi):
        print(df_pythia8.iloc[i]['phi1_ch'], df_pythia8.iloc[i]['phi2_ch'], dphi)
    dr.append(np.sqrt(dphi**2+(df_pythia8.iloc[i]['y1_ch']-df_pythia8.iloc[i]['y2_ch'])**2))
df_pythia8['dr'] = dr
```

```python
nBins=14
bounds = []
spacers = 0
for i in range(0, nBins+1):
    bounds.append(pow(10, spacers/15*(np.log10(0.4)-np.log10(0.02))+np.log10(0.02)))
    spacers += 1
    if(i==0):
        spacers += 2
    if(i==1):
        spacers += 1
    if(i==2):
        spacers += 1
```

```python
plt.xlabel('$\Delta R(leading, subleading)$')
plt.ylabel('arb. unit, normalized by bin width')
plt.hist(df_pythia8['dr'], weights=df_pythia8['w'], bins=bounds, density=True)
```

```python
plt.xlabel('$\Delta R(leading, subleading)$')
plt.ylabel('arb. unit')
total = np.sum(plt.hist(df_pythia8['dr'], weights=df_pythia8['w'], bins=bounds)[0])
weights = plt.hist(df_pythia8['dr'], weights=df_pythia8['w'], bins=bounds)[0]/total
avg_eec = 0
for i in range(0, len(eec)):
    avg_eec += weights[i]*eec[i]
print(avg_eec)
```
